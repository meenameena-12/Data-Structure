```python
Max=5
queue=[]
def enqueue():
    if len(queue)==Max:
        print("Queue overflow")
    else:
        item=int(input("Enter element:"))
        queue.append(item)
        print("Element inserted successful.")
def dequeue():
    if len(queue)==0:
        print("Queue underflow")
    else:
        print("Deleted elemnts:",queue.pop(0))
def display():
    if len(queue)==0:
        print("Queue is empty")
    else:
        print("Queue elments:",queue)
while True:
    print("\n---Queue Menu---")
    print("1.Enqueue")
    print("2.Dequeue")
    print("3.Display")
    print("4.Exit")
    
    choice = int(input("Enter your choice:"))
    
    if choice==1:
        enqueue()
    elif choice==2:
        dequeue()
    elif choice==3:
        display()
    elif choice==4:
        break
    else:
        print("Invalid choice")
        
        
```
